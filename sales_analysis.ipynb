```python
#converting the date column to datetime format
df["Order Date"] = pd.to_datetime(df["Order Date"])
df["Ship Date"] = pd.to_datetime(df["Ship Date"])
```

```python

# Check the data types
print(df[["Order Date", "Ship Date"]].dtypes)
```

```python
# Total gross revenue
total_revenue = df["Sales"].sum()

# Total profit
total_profit = df["Profit"].sum()

# Profit margin
profit_margin = (total_profit / total_revenue) * 100

# Total units sold
total_units = df["Quantity"].sum()

# Number of unique orders
total_orders = df["Order ID"].nunique()

# Average Order Value
average_order_value = total_revenue / total_orders

# Display the results
print(f"Total Gross Revenue: ${total_revenue:,.2f}")
print(f"Total Profit: ${total_profit:,.2f}")
print(f"Overall Profit Margin: {profit_margin:.2f}%")
print(f"Total Units Sold: {total_units:,}")
print(f"Total Unique Orders: {total_orders:,}")
print(f"Average Order Value: ${average_order_value:,.2f}")
```

Revenue: The dataset records approximately $2.33 million in sales.

Profit: The recorded profit is about $292,297.

Profit margin: Approximately $12.56 of profit was recorded for every $100 in sales.

Average order value: Each distinct order generated an average of $455.20 in sales.

Units sold: The recorded quantity totals 38,654 units across 5,111 distinct orders.

```python
# Group sales data by category
category_sales = df.groupby("Category").agg(
    Total_Revenue=("Sales", "sum"),
    Total_Profit=("Profit", "sum"),
    Units_Sold=("Quantity", "sum"),
    Unique_Orders=("Order ID", "nunique")
)

# Calculate profit margin
category_sales["Profit_Margin_%"] = (
    category_sales["Total_Profit"]
    / category_sales["Total_Revenue"]
) * 100

# Sort by revenue, highest first
category_sales = category_sales.sort_values(
    by="Total_Revenue",
    ascending=False
)

category_sales.round(2)
```

```python
plt.figure(figsize=(9, 5))

ax = sns.barplot(
    data=category_sales.reset_index(),
    x="Category",
    y="Total_Revenue"
)

plt.title("Total Revenue by Product Category", fontsize=14)
plt.xlabel("Product Category")
plt.ylabel("Total Revenue ($)")

# Add labels above the bars
for container in ax.containers:
    ax.bar_label(container, fmt="$%.0f", padding=3)

plt.ticklabel_format(style="plain", axis="y")
plt.tight_layout()

plt.savefig(
    "charts/revenue_by_category.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
subcategory_sales = df.groupby("Sub-Category").agg(
    Total_Revenue=("Sales", "sum"),
    Total_Profit=("Profit", "sum"),
    Units_Sold=("Quantity", "sum")
)

subcategory_sales["Profit_Margin_%"] = (
    subcategory_sales["Total_Profit"]
    / subcategory_sales["Total_Revenue"]
) * 100

subcategory_sales = subcategory_sales.sort_values(
    by="Total_Revenue",
    ascending=False
)

subcategory_sales.round(2)
```

Phones generate $331,842.64 in revenue and $45,050.83 in profit.

Binders have high unit volume at 6,071 units and a positive 15.16% margin.

Bookcases and Supplies also record negative profits, so they deserve further investigation.

```python
top5_subcategories = subcategory_sales.head(5).reset_index()

plt.figure(figsize=(10, 6))

ax = sns.barplot(
    data=top5_subcategories,
    x="Total_Revenue",
    y="Sub-Category"
)

plt.title("Top 5 Sub-Categories by Revenue")
plt.xlabel("Total Revenue ($)")
plt.ylabel("Sub-Category")

for container in ax.containers:
    ax.bar_label(container, fmt="$%.0f", padding=3)

plt.tight_layout()

plt.savefig(
    "charts/top5_subcategories_by_revenue.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
margin_data = (
    subcategory_sales
    .sort_values("Profit_Margin_%", ascending=False)
    .reset_index()
)

plt.figure(figsize=(11, 7))

ax = sns.barplot(
    data=margin_data,
    x="Profit_Margin_%",
    y="Sub-Category"
)

plt.axvline(0, color="black", linewidth=1)
plt.title("Profit Margin by Sub-Category")
plt.xlabel("Profit Margin (%)")
plt.ylabel("Sub-Category")

plt.tight_layout()
plt.savefig(
    "charts/profit_margin_by_subcategory.png",
    dpi=300,
    bbox_inches="tight"
)
plt.show()
```

 Sub-Category Performance Findings

The analysis identified Chairs as the highest revenue-generating sub-category, with sales of $335,768.25. Copiers recorded the highest total profit at $56,093.94, while Labels had the highest profit margin at 43.90%. In contrast, Tables recorded the largest loss of $17,753.21, followed by Bookcases and Supplies.

These findings suggest that management should investigate the costs, pricing, discount patterns, and sales performance of loss-making product lines. The business could also examine the factors contributing to the strong profitability of Copiers, Labels, and Paper before deciding whether to expand these product lines.

```python

# Extract the year and month from Order Date
df['Year_Month'] = df['Order Date'].dt.to_period('M')

# Calculate monthly revenue
monthly_sales = (
    df.groupby('Year_Month')['Sales']
    .sum()
    .reset_index()
)

# Rename the columns
monthly_sales.columns = ['Month', 'Total_Revenue']

# Display the monthly revenue
display(monthly_sales.head(12))
```

```python
#Creating a monthly revenue trend chart
import matplotlib.pyplot as plt

# Convert Month to string for chart labels
monthly_sales['Month'] = monthly_sales['Month'].astype(str)

# Plot monthly revenue
plt.figure(figsize=(14, 6))
plt.plot(
    monthly_sales['Month'],
    monthly_sales['Total_Revenue'],
    marker='o'
)

plt.title('Monthly Revenue Trends (2023–2026)')
plt.xlabel('Month')
plt.ylabel('Total Revenue')
plt.xticks(rotation=90)
plt.grid(True, alpha=0.3)
plt.tight_layout()

# Save the chart
plt.savefig(
    'revenue_by_month.png',
    dpi=300,
    bbox_inches='tight'
)

plt.show()
```

```python
#Calculate quarterly sales performance

# Extract year and quarter
df['Year'] = df['Order Date'].dt.year
df['Quarter'] = df['Order Date'].dt.quarter

# Calculate quarterly revenue and profit
quarterly_sales = (
    df.groupby(['Year', 'Quarter'])
    .agg(
        Total_Revenue=('Sales', 'sum'),
        Total_Profit=('Profit', 'sum'),
        Units_Sold=('Quantity', 'sum')
    )
    .reset_index()
)

# Create readable quarter labels
quarterly_sales['Quarter'] = (
    'Q' + quarterly_sales['Quarter'].astype(str)
)

quarterly_sales['Year_Quarter'] = (
    quarterly_sales['Year'].astype(str)
    + ' '
    + quarterly_sales['Quarter']
)

display(quarterly_sales)
```

```python
#Creating a quarterly revenue trend chart
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 6))

plt.plot(
    quarterly_sales['Year_Quarter'],
    quarterly_sales['Total_Revenue'],
    marker='o'
)

plt.title('Quarterly Revenue Trends (2023–2026)')
plt.xlabel('Year and Quarter')
plt.ylabel('Total Revenue')
plt.xticks(rotation=45)
plt.grid(True, alpha=0.3)
plt.tight_layout()

# Save the chart
plt.savefig(
    'quarterly_revenue.png',
    dpi=300,
    bbox_inches='tight'
)

plt.show()
```

```python
#comparing  quarterly profit
plt.figure(figsize=(12, 6))

plt.plot(
    quarterly_sales['Year_Quarter'],
    quarterly_sales['Total_Profit'],
    marker='o'
)

plt.title('Quarterly Profit Trends (2023–2026)')
plt.xlabel('Year and Quarter')
plt.ylabel('Total Profit')
plt.xticks(rotation=45)
plt.grid(True, alpha=0.3)
plt.tight_layout()

# Save the chart
plt.savefig(
    'quarterly_profit.png',
    dpi=300,
    bbox_inches='tight'
)

plt.show()
```

```python
#investigaing  seasonal fluctuations

seasonal_sales = (
    quarterly_sales.groupby('Quarter')
    .agg(
        Average_Revenue=('Total_Revenue', 'mean'),
        Average_Profit=('Total_Profit', 'mean'),
        Average_Units_Sold=('Units_Sold', 'mean')
    )
    .reset_index()
)

display(seasonal_sales.round(2))
```

Q1 is the weakest quarter on average, with the lowest revenue, profit, and sales volume.

Revenue increases progressively from Q1 through Q4.

Q4's average revenue is approximately 2.43 times Q1's average revenue.

The pattern suggests possible seasonal demand, but we should not conclude that seasonality is the cause without further investigation.

```python

plt.figure(figsize=(9, 5))

plt.bar(
    seasonal_sales['Quarter'],
    seasonal_sales['Average_Revenue']
)

plt.title('Average Revenue by Quarter (2023–2026)')
plt.xlabel('Quarter')
plt.ylabel('Average Revenue')
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
# Save the chart
plt.savefig(
    'average_revenue_by_quarter.png',
    dpi=300,
    bbox_inches='tight'
)

plt.show()
```

Seasonal Sales Performance

The quarterly analysis showed a consistent increase in average sales revenue from the first quarter to the fourth quarter. Q1 recorded the lowest average revenue of 91,623.21, while Q4 recorded the highest average revenue of 222,502.61. Average profit and units sold followed a similar pattern, with Q4 recording 28,246.40 in average profit and 3,665.75 units sold.

These findings suggest that sales performance was strongest in the fourth quarter of the available data. The business could investigate factors contributing to this pattern when planning inventory and marketing activities. However, the results show an association rather than proving the cause of the increase. The analysis also includes future-dated records in 2026, which should be considered when interpreting the findings.

```python

# Extract the month number and month name
df['Month_Number'] = df['Order Date'].dt.month
df['Month_Name'] = df['Order Date'].dt.month_name()

# Calculate average monthly revenue across the available years
monthly_seasonality = (
    df.groupby(['Month_Number', 'Month_Name'])
    .agg(
        Average_Revenue=('Sales', 'sum'),
        Total_Orders=('Order ID', 'nunique')
    )
    .reset_index()
)

# Calculate average revenue per year for each calendar month
monthly_seasonality['Average_Revenue'] = (
    monthly_seasonality['Average_Revenue'] / 4
)

# Display months in calendar order
monthly_seasonality = monthly_seasonality.sort_values('Month_Number')

display(monthly_seasonality.round(2))
```

Grouping sales by calendar month, such as January across all four years.

Counting unique orders for each month.

Dividing total monthly revenue by four to calculate average monthly revenue.

```python

plt.figure(figsize=(12, 5))

plt.plot(
    monthly_seasonality['Month_Name'],
    monthly_seasonality['Average_Revenue'],
    marker='o'
)

plt.title('Average Monthly Revenue (2023–2026)')
plt.xlabel('Month')
plt.ylabel('Average Revenue')
plt.xticks(rotation=45)
plt.grid(True, alpha=0.3)
plt.tight_layout()
# Save the chart
plt.savefig(
    'average_monthly_revenue.png',
    dpi=300,
    bbox_inches='tight'
)

plt.show()
```

```python

# Calculate revenue for each month in each year
yearly_monthly_sales = (
    df.groupby(['Year', 'Month_Number'])['Sales']
    .sum()
    .reset_index()
)

# Convert month numbers to month names
yearly_monthly_sales['Month_Name'] = (
    pd.to_datetime(
        yearly_monthly_sales['Month_Number'],
        format='%m'
    ).dt.month_name()
)

# Display the monthly revenue by year
display(
    yearly_monthly_sales.pivot(
        index='Month_Name',
        columns='Year',
        values='Sales'
    ).reindex([
        'January', 'February', 'March', 'April',
        'May', 'June', 'July', 'August',
        'September', 'October', 'November', 'December'
    ]).round(2)
)
```

```python

# Prepare monthly revenue for plotting
monthly_pivot = yearly_monthly_sales.pivot(
    index='Month_Number',
    columns='Year',
    values='Sales'
).sort_index()

# Use month names on the x-axis
month_labels = pd.to_datetime(
    monthly_pivot.index,
    format='%m'
).month_name()

plt.figure(figsize=(13, 6))

for year in monthly_pivot.columns:
    plt.plot(
        month_labels,
        monthly_pivot[year],
        marker='o',
        label=str(year)
    )

plt.title('Monthly Revenue Comparison by Year')
plt.xlabel('Month')
plt.ylabel('Total Revenue')
plt.xticks(rotation=45)
plt.legend(title='Year')
plt.grid(True, alpha=0.3)
plt.tight_layout()

# Save the chart
plt.savefig(
    'monthly_revenue_comparison.png',
    dpi=300,
    bbox_inches='tight'
)

plt.show()
```

```python

# Create a copy of monthly sales
monthly_growth = monthly_sales.copy()

# Convert Month back to a monthly period
monthly_growth['Month'] = pd.PeriodIndex(
    monthly_growth['Month'], freq='M'
)

# Calculate the percentage change from the previous month
monthly_growth['Revenue_Growth_%'] = (
    monthly_growth['Total_Revenue']
    .pct_change() * 100
)

# Display monthly revenue and growth
display(monthly_growth.round(2))
```

```python

plt.figure(figsize=(14, 5))

plt.plot(
    monthly_growth['Month'].astype(str),
    monthly_growth['Revenue_Growth_%'],
    marker='o'
)

plt.axhline(y=0, linestyle='--')

plt.title('Month-over-Month Revenue Growth (2023–2026)')
plt.xlabel('Month')
plt.ylabel('Revenue Growth (%)')
plt.xticks(rotation=90)
plt.grid(True, alpha=0.3)
plt.tight_layout()

# Save the chart
plt.savefig(
    'monthly_revenue_growth.png',
    dpi=300,
    bbox_inches='tight'
)

plt.show()
```

```python

# Five months with the highest revenue
top_months = monthly_growth.nlargest(5, 'Total_Revenue')

# Five months with the lowest revenue
bottom_months = monthly_growth.nsmallest(5, 'Total_Revenue')

print("TOP 5 MONTHS BY REVENUE")
display(top_months.round(2))

print("BOTTOM 5 MONTHS BY REVENUE")
display(bottom_months.round(2))
```
