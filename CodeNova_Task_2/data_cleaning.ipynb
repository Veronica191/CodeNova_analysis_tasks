```python

# 1. Convert date columns to datetime
date_columns = ["Order Date", "Ship Date"]

for col in date_columns:
    df[col] = pd.to_datetime(df[col], errors="coerce")

# 2. Convert identifier columns to strings
identifier_columns = ["Order ID", "Customer ID", "Product ID"]

for col in identifier_columns:
    df[col] = df[col].astype("str")

# 3. Convert categorical columns to category
categorical_columns = [
    "Ship Mode",
    "Segment",
    "Country/Region",
    "City",
    "State/Province",
    "Region",
    "Category",
    "Sub-Category"
]

for col in categorical_columns:
    df[col] = df[col].astype("category")

# 4. Keep Postal Code as a string
df["Postal Code"] = df["Postal Code"].astype("str")

# 5. Verify the new data types
print("--- UPDATED DATA TYPES ---")
print(df.dtypes)

print("\n--- CHECK DATE CONVERSION ---")
print("Invalid Order Dates:", df["Order Date"].isna().sum())
print("Invalid Ship Dates:", df["Ship Date"].isna().sum())

print("\n--- UPDATED DATASET SHAPE ---")
print(df.shape)
```

```python

# Check for inconsistent column names
print("--- COLUMN NAMES ---")
print(df.columns.tolist())

# Check for columns with only one unique value
print("\n--- CONSTANT COLUMNS ---")
for col in df.columns:
    if df[col].nunique(dropna=False) <= 1:
        print(col, "has only one distinct value")

# Check for negative values in numerical columns
print("\n--- NEGATIVE VALUES ---")
numeric_df = df.select_dtypes(include="number")

for col in numeric_df.columns:
    negative_count = (numeric_df[col] < 0).sum()
    if negative_count > 0:
        print(f"{col}: {negative_count} negative values")

# Check for columns that may contain inconsistent text
print("\n--- TEXT VALUES WITH EXTRA SPACES ---")
for col in df.select_dtypes(include="object").columns:
    text_values = df[col].dropna().astype(str)
    extra_spaces = text_values.ne(text_values.str.strip()).sum()

    if extra_spaces > 0:
        print(f"{col}: {extra_spaces} values with leading/trailing spaces")
```

```python

# Detect potential outliers using IQR
numeric_df = df.select_dtypes(include="number")

outlier_results = []

for col in numeric_df.columns:
    series = numeric_df[col].dropna()

    if series.empty:
        continue

    q1 = series.quantile(0.25)
    q3 = series.quantile(0.75)
    iqr = q3 - q1

    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr

    outliers = series[
        (series < lower_bound) | (series > upper_bound)
    ]

    outlier_results.append({
        "Column": col,
        "Lower_Bound": lower_bound,
        "Upper_Bound": upper_bound,
        "Outlier_Count": len(outliers),
        "Outlier_Percentage": round(
            len(outliers) / len(series) * 100, 2
        )
    })

outlier_summary = pd.DataFrame(outlier_results)

print(outlier_summary.to_string(index=False))
```

```python
# Inspect the highest potential outliers
for column in ["Sales", "Quantity", "Discount", "Profit"]:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower_bound) |
        (df[column] > upper_bound)
    ]

    print(f"\n--- {column} ---")
    print(f"Upper bound: {upper_bound:.2f}")
    print(outliers.sort_values(
        by=column, ascending=False
    )[[column]].head(5))
```

```python
# Calculate the IQR upper bound for Sales
Q1 = df["Sales"].quantile(0.25)
Q3 = df["Sales"].quantile(0.75)
IQR = Q3 - Q1

upper_bound = Q3 + 1.5 * IQR

# Select potential sales outliers
sales_outliers = df[df["Sales"] > upper_bound]

# Inspect the 10 highest sales values
columns_to_view = [
    "Order ID",
    "Product Name",
    "Sales",
    "Quantity",
    "Discount",
    "Profit"
]

print(sales_outliers[columns_to_view]
      .sort_values(by="Sales", ascending=False)
      .head(10)
      .to_string(index=False))
```

```python
# Calculate IQR bounds for Profit
Q1 = df["Profit"].quantile(0.25)
Q3 = df["Profit"].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Select potential profit outliers
profit_outliers = df[
    (df["Profit"] < lower_bound) |
    (df["Profit"] > upper_bound)
]

# Inspect the 10 largest losses
columns_to_view = [
    "Order ID",
    "Product Name",
    "Sales",
    "Discount",
    "Profit"
]

print("Lower bound:", lower_bound)
print("\n10 largest losses among potential outliers:")

print(
    profit_outliers[profit_outliers["Profit"] < lower_bound]
    [columns_to_view]
    .sort_values(by="Profit")
    .head(10)
    .to_string(index=False)
)
```
